<br>


```python
import sympy as sp
#fedinimos la matrix de esfuerzos
sigma_1=sp.symbols("sigma_1")
sigma_2=sp.symbols("sigma_2")
sigma_3=sp.symbols("sigma_3")

print("partimos del sistema de esfuerzos:")
Tensor=sp.Matrix([
    [sigma_1,0,0],
    [0,sigma_2,0],
    [0,0,sigma_3]
])
display(Tensor)
print("---------------------------------------------------------------------------")
print("ahora hallamos los cosenos directores de la base: ")
print("como cada angulo de inclinacion es el mismo entonces la magnitud de las componentes de n gorrito deben ser iguales")
alfa=sp.symbols("alpha")
beta=sp.symbols("beta")
gama=sp.symbols("gamma")
n_gorrito=sp.Matrix([alfa,beta,gama])
display(n_gorrito)
print("la magnitud de alfa,beta y gamma son iguales, y como es un vector unitario por geometria vectorial : ")
suma1=sp.Eq((n_gorrito[0]**2)+(n_gorrito[1]**2)+(n_gorrito[2]**2),1)
display(suma1)
print("esta suma da 1")
print("ahora como sabemos que la magnitud es igual, reemplazamos y despejamos:")
suma2=sp.Eq((n_gorrito[0]**2+n_gorrito[0]**2+n_gorrito[0]**2),1)
display(suma2)
print("como habiamos dicho que esa suma daba uno, despejo alfa da: ")
solucion1=sp.solve(suma2,n_gorrito[0])
display(solucion1)
display(sp.Eq(solucion1[0],alfa),sp.Eq(solucion1[1],alfa))
print("por lo cual beta y gama tanbien seran iguales a alfa dandonos 8 posibles combinaciones (2*2*2)")
print("---------------------------------------------------------------------------")
print("por ahora usaremos los cosenos directores positivos, aplicaremos la formula ce cauchy:")
esfuerzo_octaedro_normal=sp.symbols("sigma_OCT")
n=sp.Matrix([solucion1[1],solucion1[1],solucion1[1]])
print("definimos q")
q=Tensor*n
display(q)
print("usando cauchu q*n: ")



esfuerzo_oct=q.dot(n) ##producto punto da un escalar


##para mostrarlo bonoito en pantalla
display(sp.Eq(esfuerzo_oct,esfuerzo_octaedro_normal)) ##para verlo como eacuacion
print("este es el esfuerzo octaedrico, la componenet normal del vector q, ahora se hallara la componente cortante del vector q")
print("---------------------------------------------------------------------------")
print("ahora hallaremos el esfuerzo octaedrico cortante al descomponer q usando pitagoras")
display(sp.Eq(sp.symbols("q")**2,(sp.symbols("tau_OCT")**2+esfuerzo_octaedro_normal**2)))##para verlo visual
print("despejamos:")
display(sp.Eq(sp.symbols("q")**2-esfuerzo_octaedro_normal**2,(sp.symbols("tau_OCT")**2)))
print("reemplazo y opero")
ecuacion3=sp.Eq(sp.sqrt((q[0]**2+q[1]**2+q[2]**2)-esfuerzo_oct**2),sp.symbols("tau_OCT"))
display(ecuacion3)
print("simplifico: ")


Toct=sp.simplify(sp.sqrt((q[0]**2+q[1]**2+q[2]**2)-esfuerzo_oct**2)) #esfuerzo octaedrico


##para mostrarlo bonoito en pantalla
display(sp.Eq(Toct,sp.symbols("tau_OCT")))
print("esta expresion es equivalente a la del libro:")
```
