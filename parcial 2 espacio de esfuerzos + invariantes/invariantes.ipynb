```python
import sympy as sp
import numpy as np

## definimos las variables
sigmax=sp.symbols("sigma_x")
sigmay=sp.symbols("sigma_y")
sigmaz=sp.symbols("sigma_z")
tauxy=sp.symbols("tau_xy")
tauxz=sp.symbols("tau_xz")
tauyz=sp.symbols("tau_yz")


matrizEsfuerzos=sp.Matrix([
    [sigmax,tauxy,tauxz],
    [tauxy,sigmay,tauyz],
    [tauxz,tauyz,sigmaz]])

display(matrizEsfuerzos)

#defino la amtriz identidad

identidad=sp.Matrix([
    [1,0,0],
    [0,1,0],
    [0,0,1]
])
display(identidad)

#defino un simbolo algebraico para los esfuerzos principales usare sigma n
esfuerzos_principales=sp.symbols("sigma_n")
display(esfuerzos_principales)

#defino la ecuacion 2
ecuacion1=(matrizEsfuerzos-(esfuerzos_principales*identidad))
display(ecuacion1)  ## esta es la matriz de la ecuacion 2


determinante=ecuacion1.det()
display(determinante)

#lo agrupo con respecto a sigma n
determinante_simplificado=sp.collect(determinante, esfuerzos_principales)
display(determinante_simplificado)

#ya simplificado almacenamos esos invariantes en I1,i2,i3  teneindo en cuenta los signos del libro
print("invariante1")
invariante1=determinante_simplificado.coeff(esfuerzos_principales, 2)  #va de derecha a izquierda empezando desde el cero
display(invariante1)
print("invariante2")
invariante2=-(determinante_simplificado.coeff(esfuerzos_principales,1))
display(invariante2)
print("invariante3")
invariante3=determinante_simplificado.coeff(esfuerzos_principales,0)
display(invariante3)
```
