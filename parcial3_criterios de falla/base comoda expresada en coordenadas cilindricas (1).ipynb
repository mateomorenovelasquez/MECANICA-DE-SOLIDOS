```python
import sympy as sp
#vamos a definir ro
print("primero vamos a definir ro: ")
ro=sp.symbols("rho")
display(ro)

print("ro representara la coordenada radial es decir la distancia de OP y como esta en el plano pi representara sus componentes deviadora")
desv_1=sp.symbols("S_1")
desv_2=sp.symbols("S_2")
desv_3=sp.symbols("S_3")
desviadores=sp.Matrix([desv_1,desv_2,desv_3])

## ecuacion para verlo visual
display(sp.Eq(ro,sp.symbols("OP")))
display(sp.Eq(ro,desviadores, evaluate=False)) # el evaluate false es para que no salga falso y salga la ecuacion
print("tanbien podemos ver al punto p como un punto con coordenadas (-sigmaE,sigmaF)")
print("con lo cual podemos hallar esta dsitancia de forma pitagorica")
sigmaf=sp.symbols("sigma_F")
sigmae=sp.symbols("sigma_E")
display(sp.Eq(ro,sp.sqrt(sigmae**2+sigmaf**2)))

print("recordemos que sigmaE y sigmaF es igual a:")
sigma_e=(1/sp.sqrt(2))*(sp.symbols("sigma_3")-sp.symbols("sigma_2"))
sigma_f=(1/sp.sqrt(6))*(2*(sp.symbols("sigma_1"))-sp.symbols("sigma_2")-sp.symbols("sigma_3"))

#verlo visualmente
display(sp.Eq(sigmae,sigma_e),sp.Eq(sigmaf,sigma_f))

print("al reemplazar y simplificar obtenemos:")
ro_resultado_sin_invariante_j2=sp.simplify(sp.sqrt(sigma_e**2+sigma_f**2))
display(sp.Eq(ro,ro_resultado_sin_invariante_j2))

print("esta es una manera de escribir ro, ahora veremos si es igual a la del libro definiendo j y compararandolos con .equals()")
j2_simbolico=sp.symbols("J_2")
j2=((sp.symbols("sigma_1")-sp.symbols("sigma_2"))**2+(sp.symbols("sigma_2")-sp.symbols("sigma_3"))**2+(sp.symbols("sigma_3")-sp.symbols("sigma_1"))**2)/6
display(sp.Eq(j2_simbolico,j2))
print("ro del libro con j2:")
ro_libro=sp.sqrt(2*j2)
display(sp.Eq(ro,sp.sqrt(2*j2_simbolico)))

print("comparamos y nos da:")
comparacion_de_ro=ro_resultado_sin_invariante_j2.equals(ro_libro)
display(comparacion_de_ro)

print("dio verdadero por lo cual ambas expresiones son iguales")
print("ro mide la intensidad del esfuerzo desviador  (que tan grande es el esfuerzo para que el maerial se distorcione)")
print("------------------------------------------------------------------------------------------------")
```

```python
print("ahora definiremos el angulo de lode que va de o a 60 grados, es el angulo entre el vector unitario nf y la linea OP")
print("para definirlo usaremos la ecuacion del angulo entre dos vectores que esta arriba")
print("definimos el vector nf (ya hallado en otro codigo) y el vector OP (en sus componentes desviadores)")
print("nf;")
nf_simbolico=sp.symbols("n_F")
nf=sp.Matrix([(1/sp.sqrt(6))*2,(1/sp.sqrt(6))*-1,(1/sp.sqrt(6))*-1])
display(sp.Eq(nf_simbolico,nf,evaluate=False))
print("OP (lo escirbi arriba): ")
display(sp.Eq(ro,desviadores, evaluate=False))
print("ahora hallare la norma de cada vector: ")



##funcion que calcula la norma
funcion_norma= lambda vector: sp.sqrt(vector[0]**2+vector[1]**2+vector[2]**2)




norma_nf=funcion_norma(nf)
norma_ro=sp.sqrt(2*j2_simbolico)

display(sp.Eq(ro,norma_ro))
display(sp.Eq(nf_simbolico,norma_nf))
##############################################################################################################
print("reemplazo y obtenemos que: (forma 1 con p)")
#para verlo bonito:
teta=sp.symbols("theta")
costeta=sp.cos(teta)
##############################################################################################################
cosTeta=sp.simplify(nf.dot(desviadores)/(norma_nf*ro))
display(sp.Eq(costeta,cosTeta))
print("reemplazo y obtenemos que: (forma 1 con j2)")
display(sp.Eq(costeta,cosTeta.subs(ro,sp.sqrt(2*j2_simbolico))))
###############################################################################################
print("recordemos que j1 es igual a 0 lo que es igual a:")
j1=sp.symbols("J_1")
display(sp.Eq(j1,desv_1+desv_2+desv_3))
display(sp.Eq(desv_1+desv_2+desv_3,0))

print("esto devido a que no tiene aportacion volumetrica (deducido en otro codigo)")
print("por lo cual")
display(sp.Eq(desv_2,-desv_1-desv_3))
print("al reemplazarlo en la forma 1 obtenemos que:")
cosTeta_con_s1=cosTeta.subs(desv_2,-desv_1-desv_3)
cosTeta_con_s1_y_j2=cosTeta_con_s1.subs(ro,sp.sqrt(2*j2_simbolico))
display(sp.Eq(costeta,cosTeta_con_s1_y_j2))

print("esta es la forma 2 de escribir el angulo de lode en terminos del esfuerzo desviador 1 y el invariante j2 (forma 2)")
##########################################################################################################
print("ahora usando la identidad trigonometrica que esta al inicio: ")
cos3teta=sp.cos(3*teta)
display(sp.Eq(cos3teta,(4*(costeta)**3)-3*costeta))

print("reemplazo costeta de la forma 2 aqui, simplificamos y hallamos: ")
cos3Teta=sp.simplify((4*(cosTeta_con_s1_y_j2)**3)-3*cosTeta_con_s1_y_j2)
display(sp.Eq(cos3teta,cos3Teta))

print("ahora bien recordemos que: ")
j2_con_desviadores=-(desv_1*desv_2)-(desv_2*desv_3)-(desv_1*desv_3) ####################################################
display(sp.Eq(j2_simbolico,j2_con_desviadores))
display(sp.Eq(desv_1,-desv_2-desv_3))
display(sp.Eq(desv_1+desv_2+desv_3,0))

print("al reemplazar eso en el termino :")
termino=(desv_1**2-j2_simbolico)
display(termino)

print("obtenemos que")

print("1. reemplazamos j2 y le metemos ")
termino_reemplazado_j2=(termino.subs(j2_simbolico,j2_con_desviadores))
display(sp.Eq(termino,termino_reemplazado_j2))

print("2. agrupo los termino s1")
termino_reemplazado_j2_agrupado_s1=(sp.collect(termino_reemplazado_j2,desv_1))
display(sp.Eq(termino,termino_reemplazado_j2_agrupado_s1))

print("este termino es igual a:")
termino_2=desv_1*(desv_1+desv_2+desv_3)+(desv_2*desv_3)
display(sp.Eq(termino_reemplazado_j2_agrupado_s1,termino_2))
print("comprobamos: ")
termino_2.equals(termino_reemplazado_j2_agrupado_s1)
print("por lo cual: ")
display(sp.Eq(termino,termino_2))

print("ahora como dijimos que s1+s2+s3=0 entonces al treemplazar:")
termino_definitivo=termino_2.subs(desv_1+desv_2+desv_3,0)
display(sp.Eq(termino,termino_definitivo))
print("ahora como ese termino esta siendo multiplicado por s1: ")
display(sp.Eq(desv_1*termino,desv_1*termino_definitivo))

print("reemplazamos en la forma 2 para describir el angulo de lode")
cos3Teta_con_s1_s2_s3_y_j2=cos3Teta.subs(desv_1*termino,desv_1*termino_definitivo)
display(sp.Eq(cos3teta,cos3Teta_con_s1_s2_s3_y_j2))

print("ahora como sabemos que:")
j3_simbolico=sp.symbols("J_3")
display(sp.Eq(j3_simbolico,desv_1*desv_2*desv_3))

print("entonces: ")
cos3teta_j3_j2=cos3Teta_con_s1_s2_s3_y_j2.subs(desv_1*termino_definitivo,j3_simbolico)
display(sp.Eq(cos3teta,cos3teta_j3_j2))
print("esta es la forma 3 de escribir el angulo de lode en funcion de los invariantes j2 y j3")
print("el angulo de lode me indica el tipo de esfuerzo desviador que domina en el material (traccion,compresion,cortante puro, o una mezcla)")

print("-----------------------------------------------------------------------------------------------------------")
```
